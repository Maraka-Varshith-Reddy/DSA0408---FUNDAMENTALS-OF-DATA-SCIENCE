```python
import numpy as np

house_data = np.genfromtxt("house_data.csv", delimiter=",", skip_header=1, filling_values=np.nan)

print("House Data")
print("-" * 50)
print(f"{'Bedrooms':<12}{'Square Footage':<20}{'Sale Price':<15}")
print("-" * 50)

for row in house_data:
    print(f"{int(row[0]):<12}{int(row[1]):<20}{int(row[2]):<15}")

house_data = house_data[np.all(np.isfinite(house_data), axis=1)]
house_data = house_data[np.all(house_data >= 0, axis=1)]

bedrooms = house_data[:, 0]
sale_prices = house_data[:, 2]

filtered_prices = sale_prices[bedrooms > 4]

print("\nPreprocessing Completed")
print("Total houses:", len(house_data))
print("Houses with more than four bedrooms:", len(filtered_prices))

if len(filtered_prices) > 0:
    average_price = np.mean(filtered_prices)
    print("Average sale price of houses with more than four bedrooms:", average_price)
else:
    print("No houses found with more than four bedrooms.")
```
