```python
import numpy as np

fuel_efficiency = np.array([25, 30, 28, 35, 32, 40, 22, 27, 38, 45, 33, 36, 29, 42, 31])

average_efficiency = np.mean(fuel_efficiency)

car1 = fuel_efficiency[0]
car2 = fuel_efficiency[5]

percentage_improvement = ((car2 - car1) / car1) * 100

print("Fuel Efficiency Data:", fuel_efficiency)
print("Average Fuel Efficiency:", round(average_efficiency, 2), "MPG")
print("Car 1 Fuel Efficiency:", car1, "MPG")
print("Car 2 Fuel Efficiency:", car2, "MPG")
print("Percentage Improvement:", round(percentage_improvement, 2), "%")
```
